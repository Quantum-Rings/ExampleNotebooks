## 2. Build and Execute a Circuit

```python
from qrisp import QuantumVariable, h, cx
from qrisp_openquantum import OpenQuantumBackend

oq = OpenQuantumBackend(
    backend="ionq:forte-1",
    shots=100,
    client_id=CLIENT_ID,
    client_secret=CLIENT_SECRET,
)

qv = QuantumVariable(2)
h(qv[0])
cx(qv[0], qv[1])

print(qv.get_measurement(backend=oq, shots=100))
# {'00': 0.5, '11': 0.5}
```
